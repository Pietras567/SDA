<b>Zadanie domowe 3</b><br>
Utwórz mapę interaktywną ilustrującą miejsca obserwacji kruka zwyczajnego w Finlandii w roku 2021.


<b>1.</b> Wczytaj dane z pliku observations.shp.

```python
import geopandas as gpd

observations = gpd.read_file('observations.shp')
```

<b>2.</b> Przeprowadź projekcję danych na EPSG:4326.

```python
observations = observations.set_crs(epsg = 4326)
```

<b>3.</b> Nanieś na pierwszą warstwę znaczniki posiadające etykietkę narzędzi z datą obserwacji.

```python
import folium
mapa = folium.Map(location = [60.25, 24.8])
folium.GeoJson(
    observations,
    tooltip = folium.GeoJsonTooltip(
        fields = ['date'],
        aliases = ['Data'],
    )
    ).add_to(mapa)
```

<b>4.</b> Nanieś na drugą warstwę mapę ciepła uwzględniającą liczbę zaobserwowanych ptaków.

```python
from folium import plugins
heat_data = [
    [row.geometry.y, row.geometry.x, row["amount"]]
    for _, row in observations.iterrows()
]

plugins.HeatMap(
    heat_data,
    name="Mapa ciepła",
).add_to(mapa)
```

<b>5.</b> Nanieś na trzecią warstwę zgrupowane znaczniki posiadające opis z rodzajem obserwatora oraz liczbą zaobserwowanych ptaków.

```python
from folium.plugins import MarkerCluster
cluster = MarkerCluster(name='Obserwator wraz z opisami').add_to(mapa)

for _, row in observations.iterrows():
    folium.Marker(
        location = [row.geometry.y, row.geometry.x],
        popup = "Rodzaj obserwatora: "
        + str(row['observer'])
        + "<br>"
        + " Liczba zaobserwowanych ptaków: "
        + str(row['amount'])
    ).add_to(cluster)
```

<b>6.</b> Na trzeciej warstwie zastosuj różne kolory ikon w zależności od rodzaju obserwatora.

```python
cluster = MarkerCluster(name='Obserwator wraz z opisami').add_to(mapa)

observer_colors = {
    'Hobbyist': 'green',
    'Professional': 'red',
    'Amateur': 'pink'
}

for _, row in observations.iterrows():
    folium.Marker(
        location = [row.geometry.y, row.geometry.x],
        popup = "Rodzaj obserwatora: "
        + str(row['observer'])
        + "<br>"
        + " Liczba zaobserwowanych ptaków: "
        + str(row['amount']),
        icon = folium.Icon(color = observer_colors[row['observer']])
    ).add_to(cluster)

```

<b>7.</b> Na trzeciej warstwie zastosuj różne ikony w zależności od tego, czy został zaobserwowany jeden bądź więcej ptaków.

```python
cluster = MarkerCluster(name='Obserwator wraz z opisami').add_to(mapa)

observer_colors = {
    'Hobbyist': 'green',
    'Professional': 'red',
    'Amateur': 'pink'
}

for _, row in observations.iterrows():
    
    if row['amount'] == 1:
        bird_icon = "fa-solid fa-feather"
    else:
        bird_icon = "fa-solid fa-crow"
        
    obs_color = observer_colors[row['observer']]
    
    folium.Marker(
        location = [row.geometry.y, row.geometry.x],
        popup = "Rodzaj obserwatora: "
        + str(row['observer'])
        + "<br>"
        + " Liczba zaobserwowanych ptaków: "
        + str(row['amount']),
        icon = folium.Icon(color = obs_color, icon = bird_icon, prefix="fa")
    ).add_to(cluster)

```
