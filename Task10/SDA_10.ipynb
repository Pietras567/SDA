<b>Ćwiczenie 1.</b> Utwórz interaktywną mapę na podstawie współrzędnych (60.25, 24.8).

```python
import folium
from folium.plugins import MarkerCluster

### your code ###
mapa = folium.Map(location = [60.25, 24.8])
```

<b>Ćwiczenie 2.</b> Nanieś na interaktywną mapę znacznik o współrzędnych (60.20426, 24.96179).

```python
### your code ###
folium.Marker([60.20426, 24.96179]).add_to(mapa)
```

<b>Ćwiczenie 3.</b> Ustaw opcję, aby po kliknięciu na znacznik pojawiał się tekst "Kumpula Campus".

```python
### your code ###
folium.Marker([60.20426, 24.96179], popup = "Kumpula Campus").add_to(mapa)
```

<b>Ćwiczenie 4.</b> Zmień znacznik na zieloną ikonę OK.

```python
### your code ###
folium.Marker([60.20426, 24.96179], popup = "Kumpula Campus", icon = folium.Icon(color = 'green', icon='ok')).add_to(mapa)
```

<b>Ćwiczenie 5.</b> Wczytaj dane z pliku ``addresses.shp`` do obiektu typu GeoDataFrame o nazwie points.

```python
import geopandas as gpd

### your code ###
points = gpd.read_file('addresses.shp')
```

<b>Ćwiczenie 6.</b> Nanieś na interaktywną mapę o stylu cartodbpositron punkty ze zmiennej points. Przypisz im nazwę "Public transport stations".

```python
### your code ###
for idx, row in points.iterrows():
    lat = row.geometry.y
    lon = row.geometry.x
    folium.Marker(location=[lat, lon], popup = "Public transport stations", 
                  icon = folium.Icon(color = 'blue')).add_to(mapa)
#folium.GeoJson(points, name = "Public transport stations").add_to(mapa)
```

<b>Ćwiczenie 7.</b> Dodaj do mapy możliwość kontroli warstw.

```python
### your code ###
#folium.TileLayer('')
folium.LayerControl().add_to(mapa)
```

<b>Ćwiczenie 8.</b> Wczytaj do zmiennej locations pary współrzędnych (x, y) ze zmiennej points.

```python
### your code ###
locations = [(row.geometry.y, row.geometry.x) for idx, row in points.iterrows()]
```

<b>Ćwiczenie 9.</b> Nanieś na interaktywną mapę o stylu cartodbpositron zgrupowane znaczniki o współrzędnych ze zmiennej locations.

```python
### your code ###
mapa = folium.Map(tiles='cartodbpositron')

cluster = MarkerCluster(name='Locations').add_to(mapa)
for l in locations:
    folium.Marker(location=[l[0], l[1]], icon = folium.Icon(color = 'red')).add_to(cluster)
folium.LayerControl().add_to(mapa)
mapa.save("mapa.html")
```
